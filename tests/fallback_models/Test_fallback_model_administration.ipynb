# Test of assigning a model_list

```python
from duckdb_assistant import DuckDBAssistant
```

```python
dda = DuckDBAssistant()
```

#### `model_list` contains all LLMs assigned to the assistant. If none were assigned through environment variables at the time of initialisation, then a default model is assigned

```python
dda.model_list
```

### Test whether it works

```python
dda.generate("Generate simple SQL to calculate correlation between two variables", explain_results=True)
```

## Create a new `DuckDBAssistant` and add a (different) list of models

```python
dda2 = DuckDBAssistant()
dda2.set_model_list(model_list = ["gemini-3.5-flash"])
```

```python
dda2.model_list
```

```python
dda2.generate(explain_results = True, prompt = "Generate simple SQL to calculate correlation between two variables")
```

#### Add one more model to the list

```python
dda2.add_models(model_list = "gemini-3.7-flash")
```

```python
dda2.model_list
```

#### Remove a model

```python
dda2.remove_models("gemini-3.7-flash")
```

```python
dda2.model_list
```

#### Add a list of models

```python
dda2.add_models(["gemini-3.6-flash", "gemini-3.7-flash"])
```

```python
dda2.model_list
```
