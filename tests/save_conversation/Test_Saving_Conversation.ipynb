## Tests
### Test 1 : Generate without any reference to saving a conversation

```python
dda.generate("write me a prompt for creating a customer table")
```

### Test 2 : Turrn save_conversation on by default at the time of initialisation

```python
dda = DuckDBAssistant(save_conversation = True)
```

#### In this case, look for a folder marked Interaction-{uuid} at the same level as this notebook where {uuid} refers to the ID of the DuckDBAssistant.

```python
dda.generate("write me another random DuckDB SQL")
```

### Test 3: Injecting a path to a conversation history folder while initialising a DuckDB Assistant

```python
dda = DuckDBAssistant(save_conversation = True, conversation_path = "Interaction-ss")
```

```python
dda.generate("write me a random DuckDB SQL whicch calls snapshot as of 01-JUL 2026 in a catalog table", use_rag = False)
```

```python
prompt  = """
Write a DuckDB SQL to import all columns from a parquet file and save to another parquet file with ZSTD compression
"""

dda.generate(prompt,
             conversation_path = "Interaction-ss-test-4"
            )
```
