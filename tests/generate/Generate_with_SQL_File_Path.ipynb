```python
from duckdb_assistant import DuckDBAssistant
import os
```

```python
dda = DuckDBAssistant()
```

```python
prompt = "CREATE TABLE GREETING AS SELECT 'HEllo World' as greeting;"
```

```python
dda.generate(prompt,sql_file_path=os.path.join(os.getcwd(),"example_generate_sql_path.sql"))
```

```python
prompt = "CREATE TABLE GREETING AS SELECT 'HEllo World, without a file' as greeting;"
dda.generate(prompt)
```
