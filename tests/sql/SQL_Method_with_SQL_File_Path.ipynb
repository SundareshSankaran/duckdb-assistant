```python
from duckdb_assistant import DuckDBAssistant
import os
```

```python
dda = DuckDBAssistant()
```

```python
resp = dda.sql(sql_file_path=os.path.join(os.getcwd(),"..","generate","example_generate_sql_path.sql"))
```

```python
prompt = "CREATE TABLE GREETING_SQL_FILE AS SELECT 'HEllo World, with a file, sql-ed' as greeting;"
dda.sql(prompt, sql_file_path = os.path.join(os.getcwd(),"example_sql_file.sql"))
```

```python
prompt = "CREATE TABLE GREETING_NO_SQL_FILE AS SELECT 'HEllo World, no file, sql-ed' as greeting;"
dda.sql(prompt)
```

```python
dda.dd.sql("SELECT * FROM GREETING_SQL_FILE")
```

```python
dda.dd.sql("SELECT * FROM GREETING_NO_SQL_FILE")
```
