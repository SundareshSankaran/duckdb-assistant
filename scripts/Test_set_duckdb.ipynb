```python
from duckdb_assistant import DuckDBAssistant
```

```python
dda = DuckDBAssistant()
```

```python
dda.dd.sql("CREATE TABLE GREETING AS SELECT 'HEllo World' as greeting;")
```

```python
import duckdb as dd
conn = dd.connect("abc.duckdb")
```

```python
dda2 = DuckDBAssistant(dd = conn)
```

```python
dda2.dd.sql("CREATE TABLE GREETING AS SELECT 'HEllo World from conn' as greeting;")
```

```python
dda2.dd.sql("SELECT * FROM GREETING")
```

```python
dda.dd.sql("SELECT * FROM GREETING")
```
