```python
import duckdb as dd
from dataclasses import dataclass
```

```python
con = dd.connect()
```

```python
con.sql("INSTALL httpfs;LOAD httpfs; ")
```

```python
sql_string = "CREATE TABLE ABC AS SELECT * from read_csv_auto('https://raw.githubusercontent.com/sassoftware/sas-viya-dmml-pipelines/refs/heads/master/data/hmeq.csv')"
```

```python
con.sql(sql_string)
```

```python
schema = con.sql("DESCRIBE ABC").fetchall()
```

```python
duck_to_py = {
    "BOOLEAN": "bool",
    "TINYINT": "int",
    "SMALLINT": "int",
    "INTEGER": "int",
    "BIGINT": "int",
    "REAL": "float",
    "DOUBLE": "float",
    "DECIMAL": "float",  # or 'Decimal'
    "VARCHAR": "str",
    "DATE": "date",
    "TIMESTAMP": "datetime",
}

lines = ["@dataclass", "class HMEQ:"]
for col_name, col_type, *rest in schema:  # assuming (column_name, column_type, ...)
    py_type = duck_to_py.get(col_type, "Any")
    lines.append(f"    {col_name}: {py_type}")

print("\n".join(lines))

namespace = {}
exec("from dataclasses import dataclass\n"+"\n".join(lines), namespace)

HMEQ = namespace["HMEQ"]

hmeq =  HMEQ(BAD=0, LOAN=0, MORTDUE=0, VALUE=0, REASON="Debtcon", JOB='Professional', YOJ=23.5, DEROG=21, DELINQ=23, CLAGE=0.23, NINQ=1, CLNO=1, DEBTINC=0.1)

print(hmeq)
```

```python
duck_to_py = {
    "BOOLEAN": "bool",
    "TINYINT": "int",
    "SMALLINT": "int",
    "INTEGER": "int",
    "BIGINT": "int",
    "REAL": "float",
    "DOUBLE": "float",
    "DECIMAL": "float",  # or 'Decimal'
    "VARCHAR": "str",
    "DATE": "date",
    "TIMESTAMP": "datetime",
}

lines = ["@dataclass", "class HMEQ:"]
for col_name, col_type, *rest in schema:  # assuming (column_name, column_type, ...)
    py_type = duck_to_py.get(col_type, "Any")
    lines.append(f"    {col_name}: List[{py_type}] = field(default_factory=list)")

print("\n".join(lines))

namespace = {}
exec("from dataclasses import dataclass,field\n"+"\n".join(lines), namespace)

HMEQ = namespace["HMEQ"]
hmeq = HMEQ()
```
